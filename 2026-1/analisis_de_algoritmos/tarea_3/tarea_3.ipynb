1. Los cuatro algoritmos para evaluar polinomios de grado n.

```
## Algoritmos para evaluar grado n ##

# Algoritmo 1
def evalua_polinomio_directa(n, x):
    c = list(range(0, n + 1))

    s = c[0]
    for i in range(1, n+1):
        xn = 1
        for j in range(1, i+1):
            xn *= x
        s += c[i] * xn

    return s

# Algoritmo 2
def evalua_polinomio_log(n, x):
    c = list(range(0, n + 1))

    s = c[0]
    for i in range(1, n+1):
        s += c[i] * potencia(x, i)

    return s

def potencia(x, j):
    if j == 0:
        return 1
    elif j % 2 == 1:
        return x * potencia(x, j-1)
    else:
        t = potencia(x, j // 2)
        return t*t

# Algoritmo 3
def evalua_polinomio_precalc(n, x):
    c = list(range(0, n + 1))
    xn = 1
    s = c[0]
    for i in range(1, n+1):
        xn *= x
        s += c[i] * xn
    
    return s

# Algoritmo 4: Horner
def horner(n, x):
    c = list(range(0, n + 1))
    s = 0
    i = n

    while(i >= 0):
        s = (s*x) + c[i]
        i -= 1
    
    return s
```

```
def trasp(a, n):
    for i in range(1, n):
        for j in range(i+1, n+1):
            t = a[i][j]
            a[i][j] = a[j][i]
```

```
# MULTIPLICACIÓN CUADRADA
def mult_cuadrada(a, b, c, n):
    for i in range(1, n+1):
        for j in range(1, n+1):
            c[i][j] = 0
            for k in range(1, n+1):
                c[i][j] = c[i][j] + a[i][k] * b[k][j]
    
    return c[i][j]
```

```
# BÚSQUEDA SECUENCIAL
def seqsearch(a, x, n):
    i = n
    a[0] = x

    while (a[i] != x):
        i -= 1
    
    return i
```
